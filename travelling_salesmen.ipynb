TRAVELLING SALESMAN PROBLEM

```python
from itertools import permutations

n=4

cost=[
    [0,10,15,20],
    [10,0,35,25],
    [15,35,0,30],
    [20,25,30,0]
]
cities=list(range(1,n))
min_cost=float('inf')
best_tour=None

for perm in permutations(cities):
  tour=[0]+list(perm)+[0]
  total_cost=0

  for i in range(len(tour)-1):
    total_cost+=cost[tour[i]][tour[i+1]]

  if total_cost<min_cost:
     min_cost=total_cost
     best_tour=tour


print("OUTPUT:\n")
print("Optimal tour:"," ->".join(str(city+1)for city in best_tour))
print("Minimum Travelling cost:",min_cost)
```

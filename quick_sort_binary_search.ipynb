```python
comparisons=0

def partition(arr,low,high):
    global comparisons
    pivot=arr[low]
    i=low+1
    j=high
    while True:
        while i<=high:
            comparisons+=1
            if arr[i]>pivot:
                break
            i+=1
        while j>=low+1:
            comparisons+=1
            if arr[j]<pivot:
                break
            j-=1
        if i>=j:
            break
        arr[i],arr[j]=arr[j],arr[i]
    arr[low],arr[j]=arr[j],arr[low]
    return j

def quick_sort(arr,low,high):
    if low<high:
        p=partition(arr,low,high)
        quick_sort(arr,low,p-1)
        quick_sort(arr,p+1,high)

n=int(input("Enter number of elements: "))
arr=list(map(int,input("Enter the elements: ").split()))
print("Array before sorting:")
print(*arr)
quick_sort(arr,0,n-1)
print("Array after sorting:")
print(*arr)
print("Number of comparisons:",comparisons)
```

```python
def first_occurrence(arr,key):
    low=0
    high=len(arr)-1
    result=-1
    while low<=high:
        mid=(low+high)//2
        if arr[mid]==key:
            result=mid
            high=mid-1
        elif arr[mid]<key:
            low=mid+1
        else:
            high=mid-1
    return result

def last_occurrence(arr,key):
    low=0
    high=len(arr)-1
    result=-1
    while low<=high:
        mid=(low+high)//2
        if arr[mid]==key:
            result=mid
            low=mid+1
        elif arr[mid]<key:
            low=mid+1
        else:
            high=mid-1
    return result

arr=list(map(int,input("Enter sorted array elements: ").split()))
key=int(input("Enter element to search: "))
first=first_occurrence(arr,key)
last=last_occurrence(arr,key)

if first==-1:
    print("Element not found")
else:
    print("First occurrence index:",first)
    print("Last occurrence index:",last)
    print("Total number of occurrences:",last-first+1)
```

Insertion Sort,Bubble Sort,Selection Sort

```python
def bubble_sort(a):
  n=len(a)
  for i in range(n-1):
    for j in range(n-i-1):
      if a[j]>a[j+1]:
        a[j],a[j+1]=a[j+1],a[j]
  return a

def selection_sort(a):
  n=len(a)
  for i in  range(n-1):
    min_index=i
    for j in range(i+1,n):
      if a[j]<a[min_index]:
        a[j],a[min_index]=a[min_index],a[j]
  return a

def insertion_sort(a):
  for i in range(1,len(a)):
    key=a[i]
    j=i-1
    while j>=0 and a[j]>key:
      a[j+1]=a[j]
      j-=1
      a[j+1]=key
  return a

n=int(input("enter the number of elements:"))
a=list(map(int,input("enter the elements:").split()))
print("original array",a)
print("Bubble Sort:",bubble_sort(a.copy()))
print("Insertion Sort:",insertion_sort(a.copy()))
print("Selection Sort:",selection_sort(a.copy()))


```

Randomised Quick Sort


```python
import random

def partition(a, low, high):
    pivot_index = random.randint(low, high)
    a[pivot_index], a[high] = a[high], a[pivot_index]
    pivot = a[high]
    i = low - 1

    for j in range(low, high):
        if a[j] <= pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    a[i + 1], a[high] = a[high], a[i + 1]
    return i + 1

def randomised_quick_sort(a, low, high):
    if low < high:
        p = partition(a, low, high)
        randomised_quick_sort(a, low, p - 1)
        randomised_quick_sort(a, p + 1, high)

n = int(input("Enter the number of elements: "))
a = list(map(int, input("Enter the elements: ").split()))

print("Original array:", a)

randomised_quick_sort(a, 0, n - 1)

print("Sorted array:", a)
```
